```python
import pandas as pd

df = pd.read_csv("heart.csv")

df.head()
```

```python
print(df.shape)
print(df.info())
print(df.describe())
```

```python
df.isnull().sum()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='target', data=df)
plt.title("Heart Disease Distribution")
plt.show()
```

```python
sns.histplot(df['age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(df.corr(), annot=True, cmap='coolwarm')

plt.title("Correlation Matrix")
plt.show()
```

```python
sns.countplot(x='cp', hue='target', data=df)
plt.title("Chest Pain Type vs Heart Disease")
plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(), annot=True, cmap='coolwarm')

plt.title("Feature Correlation Matrix")
plt.show()
```

```python
sns.boxplot(x=df['chol'])
plt.title("Cholesterol Outliers")
plt.show()
```

```python
sns.boxplot(x='target', y='age', data=df)
plt.title("Age vs Heart Disease")
plt.show()
```

```python
sns.boxplot(x='target', y='thalach', data=df)
plt.title("Max Heart Rate vs Disease")
plt.show()
```

1. People aged 40–60 are more prone to heart disease.
2. Chest pain type (cp) is a strong predictor.
3. Maximum heart rate (thalach) is lower in patients with disease.
4. Oldpeak is higher in patients with heart disease.
5. Cholesterol has outliers but weaker correlation.
6. Males are slightly more affected than females.
